```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)


# Display original image

plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Take border width from user

border_width = int(input("Enter border width in pixels: "))


# Take border color from user

border_color = int(input("Enter border color (0-255): "))


# Get image dimensions

rows, cols = img.shape



# Create new image with border

bordered_image = np.ones(
    (rows + 2*border_width,
     cols + 2*border_width),
    dtype=np.uint8
) * border_color



# Place original image in center

bordered_image[
    border_width:border_width+rows,
    border_width:border_width+cols
] = img



# Display result

plt.figure(figsize=(10,5))


plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title("Original Image")
plt.axis("off")



plt.subplot(1,2,2)
plt.imshow(bordered_image,cmap='gray')
plt.title("Image With Border")
plt.axis("off")


plt.show()
```

```python
# Convert grayscale image to binary first

threshold = np.mean(img)


binary = np.zeros(img.shape,dtype=np.uint8)


for i in range(img.shape[0]):
    for j in range(img.shape[1]):

        if img[i,j] >= threshold:
            binary[i,j] = 255
        else:
            binary[i,j] = 0



# Border input

border_width = int(input("Enter border width: "))

border_color = int(input("Enter border color (0 or 255): "))



rows,cols = binary.shape



# Create bordered binary image

binary_border = np.ones(
    (rows+2*border_width,
     cols+2*border_width),
     dtype=np.uint8
) * border_color



# Insert binary image

binary_border[
    border_width:border_width+rows,
    border_width:border_width+cols
] = binary



# Display

plt.figure(figsize=(10,4))


plt.subplot(1,2,1)
plt.imshow(binary,cmap='gray')
plt.title("Binary Image")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(binary_border,cmap='gray')
plt.title("Binary Image With Border")
plt.axis("off")


plt.show()
```
